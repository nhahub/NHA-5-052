```python
import pandas as pd

categories = {
    "phone": ["mobiles"],
    "tv": ["brands", "lg-tvs", "lg-tv", "hisense-tvs", "haier-tvs", "haier-tv",
           "fresh-tvs", "fresh-tv", "ata-tvs", "armadillo-tvs", "master-electron-tvs"],
    "ac": ["air-conditioner", "air-conditioners"],
}

rows = []
for cat, handles in categories.items():
    for h in handles:
        url = f"https://dream2000.com/collections/{h}/products.json?limit=250"
        r2 = requests.get(url, headers=headers)
        for p in r2.json()["products"]:
            v = p["variants"][0]
            rows.append({
                "category": cat,
                "product_id": p["id"],
                "title": p["title"],
                "brand": p["vendor"],
                "sku": v["sku"],
                "price": float(v["price"]),
            })
        time.sleep(1)

dream = pd.DataFrame(rows).drop_duplicates(subset=["product_id"])
print(dream["category"].value_counts())
dream.head(10)
```

```python
xprs_categories = {
    "phone": ["mobile-phones-egypt"],
    "tv": ["32-inch-tv", "43-inch-tv", "55-inch-tv", "65-inch-tv", "75-inch-tv",
           "85-inch-tv", "ultra-hd-4k-tvs", "ultra-hd-8k-tvs", "neo-qled-tvs"],
    "ac": ["air-conditioners"],
}

rows = []
for cat, handles in xprs_categories.items():
    for h in handles:
        page = 1
        while True:
            url = f"https://myxprs.com/collections/{h}/products.json?limit=250&page={page}"
            products = requests.get(url, headers=headers).json()["products"]
            if not products:
                break
            for p in products:
                v = p["variants"][0]
                rows.append({
                    "category": cat,
                    "product_id": p["id"],
                    "title": p["title"],
                    "brand": p["vendor"],
                    "sku": v["sku"],
                    "price": float(v["price"]),
                })
            print(cat, h, "page", page, "→", len(products))
            page += 1
            time.sleep(1)

xprs = pd.DataFrame(rows).drop_duplicates(subset=["product_id"])
xprs["site"] = "xprs"
print(xprs["category"].value_counts())
xprs.head(10)
```

```python
import re

brands = {
    "samsung": ["samsung", "سامسونج"], "apple": ["apple", "آبل", "ابل", "أبل", "iphone", "آيفون", "ايفون"],
    "xiaomi": ["xiaomi", "شاومي", "redmi", "poco"], "honor": ["honor", "هونر"], "oppo": ["oppo", "أوبو", "اوبو"],
    "realme": ["realme", "ريلمي"], "vivo": ["vivo", "فيفو"], "infinix": ["infinix", "إنفينكس", "انفينكس"],
    "tecno": ["tecno", "تكنو"], "nokia": ["nokia", "نوكيا"], "hmd": ["hmd"], "huawei": ["huawei", "هواوي"],
    "motorola": ["motorola", "موتورولا"], "lg": ["lg", "إل جي", "ال جي"], "sharp": ["sharp", "شارب"],
    "carrier": ["carrier", "كاريير"], "tornado": ["tornado", "تورنيدو"], "hisense": ["hisense", "هايسنس"],
    "haier": ["haier", "هاير"], "tcl": ["tcl"], "fresh": ["fresh", "فريش"], "midea": ["midea", "ميديا"],
    "unionaire": ["unionaire", "يونيون"], "toshiba": ["toshiba", "توشيبا"], "ata": ["ata"],
    "skyline": ["sky line", "skyline", "سكاي لاين"], "jac": ["jac", "جاك"], "synix": ["synix", "سينيكس"],
    "white whale": ["white whale", "وايت ويل"], "zero": ["zero", "زيرو"], "panasonic": ["panasonic", "باناسونيك"],
    "sony": ["sony", "سوني"], "philips": ["philips", "فيليبس"],
}
arabic_words = {"برو": "pro", "ماكس": "max", "بلس": "plus", "الترا": "ultra", "ألترا": "ultra",
                "ريدمي": "redmi", "بوكو": "poco", "نوت": "note", "لايت": "lite", "ميني": "mini"}
series_words = ["pro", "max", "plus", "ultra", "lite", "mini", "redmi", "poco", "note", "reno", "nova"]
skip = {"HDR10", "DVBT2", "1080P", "720P", "2160P"}

def find_brand(text):
    t = str(text).lower()
    for brand, names in brands.items():
        for n in names:
            if re.search(rf"(?<![a-z]){re.escape(n)}(?![a-z])", t):
                return brand
    return None

def phone_model(title):
    t = str(title).lower()
    for ar, en in arabic_words.items():
        t = t.replace(ar, en)
    t = re.sub(r"\d+\s*(gb|tb)\b", " ", t)
    t = re.sub(r"\b[45]g\b", " ", t)
    t = re.split(r"[,(]| - | — ", t)[0]
    words = re.findall(r"[a-z0-9]+", t)
    model = next((w for w in words if re.search(r"\d", w)), None)
    if not model:
        return None
    extra = sorted(set(w for w in words if w in series_words))
    return "-".join(extra + [model])

def tv_ac_model(title):
    codes = re.findall(r"[A-Za-z0-9\-]{4,}", str(title))
    codes = [c.upper().replace("-", "") for c in codes if re.search(r"\d", c) and re.search(r"[A-Za-z]", c)]
    codes = [c for c in codes if c not in skip and not re.fullmatch(r"\d+(INCH|HP|K|HZ|W|BTU|GB)", c)]
    if not codes:
        return None
    code = max(codes, key=len)
    return code[:-4] if code.endswith("UXEG") else code

def clean_code(brand, code):
    if brand == "carrier":
        code = re.sub(r"^(53|42|38)", "", code)
        code = re.sub(r"708F?$", "", code)
    if brand == "midea":
        m = re.match(r"(.*?\d{2}[CH]R)", code)
        if m:
            code = m.group(1)
    return code

def make_key(category, brand_text, title):
    brand = find_brand(brand_text) or find_brand(title)
    if not brand:
        return None
    if category == "phone":
        model = phone_model(title)
    else:
        model = tv_ac_model(title)
        if model:
            model = clean_code(brand, model)
    if not model:
        return None
    return f"{brand}|{category}|{model}"

# ---- العدّ ----
is_cooler = jumia["title"].str.contains("cooler|desert|liters|litres", case=False)
jumia = jumia[~((jumia["category"] == "ac") & is_cooler)]

dream["key"] = [make_key(c, b, t) for c, b, t in zip(dream["category"], dream["brand"], dream["title"])]
jumia["key"] = [make_key(c, "", t) for c, t in zip(jumia["category"], jumia["title"])]
xprs["key"] = [make_key(c, b, t) for c, b, t in zip(xprs["category"], xprs["brand"], xprs["title"])]

all_products = pd.concat([dream, jumia, xprs])
ok = all_products.dropna(subset=["key"])
sites_per_key = ok.groupby(["category", "key"])["site"].nunique().reset_index(name="n_sites")

summary = pd.DataFrame({
    "on_2plus_sites": sites_per_key[sites_per_key["n_sites"] >= 2].groupby("category").size(),
    "on_3_sites": sites_per_key[sites_per_key["n_sites"] == 3].groupby("category").size(),
}).fillna(0).astype(int)
summary["decision"] = ["keep ✅" if n >= 20 else "drop ❌" for n in summary["on_2plus_sites"]]
summary
```
